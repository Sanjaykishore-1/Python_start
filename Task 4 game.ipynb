```python
## Word guessing game 
## Mini Hangman
def Mini_Hangman():
    import random
        
    words = ["python","machine","learning","bridge","loops","logic"]
    computer_choice = random.choice(words)
    
    print("Welcome to Word guessing game >>>> Mini Hangman !!")
    
    length_of_word = len(computer_choice)
    blank_space = "_ " * len(computer_choice)
    print("lets play now")
            
    for i in range(0,6):
        try:
                user_game_input = input(f"enter your guessing word:{blank_space}").lower()
                if  user_game_input != computer_choice: 
                    print("wrong guess try again!! :(")
                elif (user_game_input).isdigit():
                    print("enter value input")
                    continue
                else:
                  print(f"your guess is correct :) !!! the word is {computer_choice}")
                  break
        except int(ValueError, KeyError):
                print("Enter valued input from choices")    
```

```python
Mini_Hangman()
```
