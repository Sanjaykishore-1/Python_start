```python

class motorcycle:

    def __init__(self,brand,model,year,engine,plate_no):
        self.brand   = brand
        self.model   = model
        self.year    = year
        self.engine  = engine
        self.plate_no = plate_no
        
    def bike_engine(self):
        print(f"the motorcycle is {self.brand}{self.model} with the engine of {self.engine}")

    def bike_info(self):
        print(f"the motorcycle is {self.brand}{self.model} with the engine of {self.engine} from the year of {self.year}") 

class Officer:    
    def check_plate(self,bike):
        print(f" as by the Officer check complete. Plate Number: {bike.plate_no}")
        

hunter = motorcycle("RE", "Hunter", 2023, "J-series", "TN11AA3903")
classic = motorcycle("RE", "Classic", 2024, "J-series", "TN07ZZ1234")

police = Officer()
```

### Abstraction

* abstraction is nothing but blueprint of the class that hiding the unnessasary implementation details and showing only the essential features of an object

```python
from abc import ABC, abstractmethod

class MotorcycleBase(ABC):

    @abstractmethod
    def bike_engine(self):
        pass

    @abstractmethod
    def bike_info(self):
        pass
```

#### Polymorphism 

* polymorphism is nothing but the same function name behave differently on object or a data\
* bike.engine() function behaves differently in every below situation.


```python
class motorcycle:
    def bike_engine(self):
        print("General motorcycle engine")

class SportBike(motorcycle):
    def bike_engine(self):
        print("High performance engine")

class CruiserBike(motorcycle):
    def bike_engine(self):
        print("Smooth cruising engine")

bikes = [motorcycle(),SportBike(), CruiserBike()]

for b in bikes:
    b.bike_engine()
```

#### Inheritance 

* Inheritance is the method of creating child class form the parent class

```python
class motorcycle:
    def bike_engine(self):
        print("General motorcycle engine")

class SportBike(motorcycle):
    def bike_engine(self):
        print("High performance engine")
```
