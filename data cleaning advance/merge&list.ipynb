## MERGE function example

```python
import pandas as pd
```

```python
strickers_scores = pd.DataFrame({"players_name" : ["sanjay","shayam","harish","sridhar","lokesh","manoj"],
                                "players_scores" : [50,40,33,27,9,21]})
print(strickers_scores)
```

```python
strickers2_scores = pd.DataFrame({"players_name" : ["vasanth","kumaran","pradeep","vicky","akash","narayanan"],
                                "players_scores" : [69,23,0,19,7,18]})
print(strickers2_scores)
```

```python
merged = strickers_scores.merge(
         strickers2_scores,how = 'inner')
print(merged)
```

```python
merged_table = pd.merge(
    strickers_scores, 
    strickers2_scores, 
    left_index=True, 
    right_index=True, 
    suffixes=('_Team1', '_Team2')
)
print(merged_table)
```

```python
merged = pd.merge(
    left=strickers_scores, 
    right=strickers2_scores, 
    left_index=True, 
    right_index=True, 
    suffixes=('_Team1', '_Team2')
)
merged  
```

```python
merged = pd.merge(
    strickers_scores, 
    strickers2_scores, 
    left_on='players_name', 
    right_on='players_scores', 
    how='inner'
)

print(merged)  ## Doubt - why output not showing with this command
```

```python
players_name = ["vasanth","kumaran","pradeep","vicky","akash","narayanan"]

filtered_players = list(filter(lambda x: "y" in x,players_name))
print(filtered_players)
```

```python
players = ["Vasanth", "Kumaran", "Pradeep", "Vicky", "Akash"]
Y_team = list(filter(lambda name: name.endswith("y"), players))
print(Y_team)
```

```python
from functools import reduce
```

```python
 ## EX 2   - Reduce 

employees_scores = {'names': ["Arun", "Santhosh", "Vijay", "Sanjay", "Deepak", 
                     "Vasanth", "Vicky", "Akash", "Rahul", "Aditya"],
                    'scores' : ["25","32","19","40","27",
                           "30","31","21","42","24"]}

int_scores = map(int, employees_scores['scores'])
total_sum = reduce(lambda x, y: x + y, int_scores)

print(total_sum)
```
