```python
from scipy import stats
import numpy as np
```

```python
12.	A dice manufacturer claims that a die is fair.
The die is rolled 120 times, and observed frequencies are recorded.
Test at the 5% level of significance whether the die is fair.


To test whether the die is fair, we can use the chi-square goodness-of-fit test. The null hypothesis (H0) is that the die is fair, 
meaning that each face has an equal probability of 1/6. The alternative hypothesis (H1) is that the die is not fair.
```

```python
observed = [20, 20, 20, 20, 20, 20] 
n = 120

expected_freq = n / 6
expected = [expected_freq] * 6

chi2_stat, p_value = stats.chisquare(observed, expected)

alpha = 0.05

if p_value < alpha:
    decision = "Reject H0: The die is not fair."
else:
    decision = "Fail to reject H0: The die is fair."

print(f"Decision: {decision}")
```
