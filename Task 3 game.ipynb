```python
# rock, paper & scissor game 
##Game module 

def rps_game():
    import random
    
    choices = {1:'rock',2:'paper',3:'scissors'}
    
    computer_choice = random.randint(1,3)
    
    print("Welcome to rock--paper--scissors game !!")
    
    while True:
        print("""lets play now
        1-Rock
        ||
        2-Paper
        ||
        3-scissors""")
        user_game_input = int(input(f"enter your choice of number(1-3):"))
        
        print(f"You chose: {choices[user_game_input]}")
        print(f"Computer chose: {choices[computer_choice]}")
        
        for i in range(0,1):
            try:
                if user_game_input == 1 and computer_choice ==2 :
                    print("ha ha ha we got you")
                elif user_game_input == 2 and computer_choice ==3:
                    print("ha ha ha we got you")
                elif user_game_input ==3 and computer_choice == 1:
                    print("ha ha ha we got you")
                elif user_game_input == 1 and computer_choice == 3:
                    print("yoouu got me!!")    
                elif user_game_input == 2 and computer_choice == 1:
                    print("yoouu got me!!")      
                elif user_game_input == 3 and computer_choice == 2:
                    print("yoouu got me!!")
                else:
                    user_game_input == computer_choice 
                    print("Its a tie")
                    break
            
            except (ValueError, KeyError):
                    print("Enter valued input from choices")   
                    continue  
                
        retry = input("wanna play again (Yes/No):").lower()
        if retry == "yes" :
            print("Thanks for playing again")
            continue
        else: retry == "no"
        print("Thank you")
        break        
```

```python
rps_game()
```
