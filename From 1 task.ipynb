```python
2.## while loop EX:1 - getting valuable mobile number

while True :
    user_input         =  input("Enter your mobile Number:")
    if  len(user_input) == 10:
        print("Thank you for the valuable input")
        break
    else:
        len(user_input) != 10
        print("Enter 10 digit mobile number")
```

```python
2.## while loop EX:2 getting valuable 2 digit from the user

while True :
    user_input     = int(input("Enter double digit number:"))
    if user_input >= 10:
       print("Thank you for the input")
       break
    else:
       user_input < 10
       print("Enter 2 digit number") 
```
